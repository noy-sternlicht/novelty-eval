```python
import os, json, tomllib, pathlib
import anthropic

with open(pathlib.Path("secrets.toml"), "rb") as f:
    secrets = tomllib.load(f)

client = anthropic.Anthropic(api_key=secrets["anthropic"])
```

```python
response = client.messages.create(
    model="claude-opus-4-6",
    max_tokens=4096,
    # thinking={"type": "adaptive", "display": "summarized"},
    output_config={"effort": "high"},
    messages=[{"role": "user", "content": "Which is tastier, dark or milk chocolate? give a short reasoning <reasoning>...</reasoning>, then your final answer (\"dark\" or \"milk\")."}],
)

print(response)

text_output = next(b.text for b in response.content if b.type == "text")
print(f"\ntext output length: {len(text_output)} chars")
```
